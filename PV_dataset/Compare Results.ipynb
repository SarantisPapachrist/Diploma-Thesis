```python
import seaborn as sns
import matplotlib.pyplot as plt
```

##### Accuracies

```python
#Random Forest
%store -r accuracy 
#Random Forest with Ranges_4 dataset 
%store -r accuracy_Ranges_4
#AdaBoost
%store -r accuracy_test 
#CN2 Rule Induction
%store -r cn2_accuracy_test 
#Logistic Regression
%store -r multi_accuracy_test 
#KNN
%store -r knn_accuracy_test 
#Naive Bayes
%store -r accuracy_test_nb 
#Support Vector Machine
%store -r svm_accuracy_test 
```

```python
classifiers = ['Random Forest', 'Random Forest (Ranges_4)', 'AdaBoost', 'CN2 Rule Induction', 'Logistic Regression', 'KNN', 'Naive Bayes', 'Support Vector Machine']
accuracies = [accuracy, accuracy_Ranges_4, accuracy_test, cn2_accuracy_test, multi_accuracy_test, knn_accuracy_test, accuracy_test_nb, svm_accuracy_test]

import pandas as pd
data = pd.DataFrame({'Classifier': classifiers, 'Accuracy': accuracies})

sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 6))
sns.barplot(x='Classifier', y='Accuracy', hue='Classifier', data=data, dodge=False)
plt.xlabel('Classifier')
plt.ylabel('Accuracy')
plt.title('Classifier Accuracies')

for index, row in data.iterrows():
    plt.text(index, row['Accuracy'] + 0.01, f'{row["Accuracy"]:.2f}', color='black', ha="center")

plt.xticks(rotation=45, ha='right')
plt.ylim(0, 1) 
plt.tight_layout()
plt.show()
```

##### Confusion Matrixes

```python
%store -r conf_matrix
%store -r conf_matrix_Ranges_4
%store -r conf_matrix_test
%store -r cn2_conf_matrix_test
%store -r multi_conf_matrix_test
%store -r knn_conf_matrix_test
%store -r conf_matrix_test_nb
%store -r svm_conf_matrix_test
```

```python
classifiers = ['Random Forest', 'Random Forest (Ranges_4)', 'AdaBoost', 'CN2 Rule Induction', 'Logistic Regression', 'KNN', 'Naive Bayes', 'Support Vector Machine']
conf_matrices = [conf_matrix, conf_matrix_Ranges_4, conf_matrix_test, cn2_conf_matrix_test, multi_conf_matrix_test, knn_conf_matrix_test, conf_matrix_test_nb, svm_conf_matrix_test]

plt.figure(figsize=(15, 10))

for i, (clf, conf_matrix) in enumerate(zip(classifiers, conf_matrices), 1):
    plt.subplot(3, 3, i)
    sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues', cbar=False)
    plt.title(clf)
    plt.xlabel('Predicted Label')
    plt.ylabel('True Label')

plt.tight_layout()
plt.show()
```
