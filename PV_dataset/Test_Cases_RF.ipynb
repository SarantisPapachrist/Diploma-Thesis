#### Load the Dataset

```python
import pandas as pd
import numpy as np

path_train = "/home/sarantis/Documents/Διπλωματική Εργασία/Diploma-Thesis/PV_dataset/datasets/train data.csv"  
train_data = pd.read_csv(path_train, delimiter=';')

path_test = "/home/sarantis/Documents/Διπλωματική Εργασία/Diploma-Thesis/PV_dataset/datasets/test data.csv"  
test_data = pd.read_csv(path_test, delimiter=';')

display(train_data)
display(test_data)
```

#### Create new Datasets without some attributes

```python
# Dataframe without Range 4
No_Range_4_train = train_data.drop('range 4', axis=1) 
No_Range_4_test = test_data.drop('range 4', axis=1) 

# Dataframe without Range 1, Range 2, Range 3, Range 4
No_Range_train = train_data.drop(['range 1', 'range 2', 'range 3', 'range 4'], axis=1)
No_Range_test = test_data.drop(['range 1', 'range 2', 'range 3', 'range 4'], axis=1)

# Dataframe μόνο με τα 4 No_Range
Ranges_4 = train_data.drop(['I1', 'I2', 'I1MAX', 'I1MIN', 'I1VAR', 'I2MAX', 'I2MIN', 'I2VAR', 'I3', 'I4', 'I3max', 'I3min', 'I3var', 'I4MAX', 'I4MIN', 'I6', 'I5', 'Itotal1', 'Itotalmax1', 'Itotalmin1', 'Vdcmean1', 'Vdcmax1', 'Vdcmin1', 'Pdcmean1', 'IR', 'T'], axis=1)
Ranges_4 = test_data.drop(['I1', 'I2', 'I1MAX', 'I1MIN', 'I1VAR', 'I2MAX', 'I2MIN', 'I2VAR', 'I3', 'I4', 'I3max', 'I3min', 'I3var', 'I4MAX', 'I4MIN', 'I6', 'I5', 'Itotal1', 'Itotalmax1', 'Itotalmin1', 'Vdcmean1', 'Vdcmax1', 'Vdcmin1', 'Pdcmean1', 'IR', 'T'], axis=1)
```

```python
# Dataframes without class
X_No_Range_4_train = No_Range_4_train.drop('class', axis=1) 
X_No_Range_4_test = No_Range_4_test.drop('class', axis=1)

X_No_Range_train = No_Range_train.drop('class', axis=1) 
X_No_Range_test = No_Range_test.drop('class', axis=1)

X_Ranges_4 = Ranges_4.drop('class', axis=1) 
X_Ranges_4 = Ranges_4.drop('class', axis=1)

# Dataframe for Target Variable
y_No_Range_4_train = No_Range_4_train['class']  
y_No_Range_4_test = No_Range_4_test['class']

y_No_Range_train = No_Range_train['class']  
y_No_Range_test = No_Range_test['class']

y_Ranges_4 = Ranges_4['class']  
y_Ranges_4 = Ranges_4['class']
```

#### Min-Max Scaling

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

X_No_Range_4_train_scaled = pd.DataFrame(scaler.fit_transform(X_No_Range_4_train), columns=X_No_Range_4_train.columns)
X_No_Range_4_test_scaled = pd.DataFrame(scaler.transform(X_No_Range_4_test), columns=X_No_Range_4_test.columns)

X_No_Range_train_scaled = pd.DataFrame(scaler.fit_transform(X_No_Range_train), columns=X_No_Range_train.columns)
X_No_Range_test_scaled = pd.DataFrame(scaler.transform(X_No_Range_test), columns=X_No_Range_test.columns)

X_Ranges_4_scaled = pd.DataFrame(scaler.fit_transform(X_Ranges_4), columns=X_Ranges_4.columns)
X_Ranges_4_scaled = pd.DataFrame(scaler.transform(X_Ranges_4), columns=X_Ranges_4.columns)
```

#### Random Forest Algorithm

```python
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestClassifier

datasets = {
    'No_Range_4': (X_No_Range_4_train_scaled, y_No_Range_4_train),
    'No_Range': (X_No_Range_train_scaled, y_No_Range_train),
    'Ranges_4': (X_Ranges_4_scaled, y_Ranges_4)
}

param_grid = {
    'max_depth': [2, 3, 5, 10, 20],
    'min_samples_leaf': [5, 10, 20, 50, 100, 200],
    'n_estimators': [10, 25, 30, 50, 100, 200]
}

# Dictionairy to save the trained models
models = {}

random_forest = RandomForestClassifier(random_state=42)

for dataset_name, (X_train, y_train) in datasets.items():
    grid_search = GridSearchCV(estimator=random_forest, param_grid=param_grid, cv=5, scoring='accuracy')
    grid_search.fit(X_train, y_train)

    best_params = grid_search.best_params_
    best_score = grid_search.best_score_

    print(f'Best Parameters for {dataset_name}: {best_params}')
    print(f'Best Mean Cross-Validated Score for {dataset_name}: {best_score}')

    random_forest_model = RandomForestClassifier(random_state=42, max_depth=best_params['max_depth'],
                                       min_samples_leaf=best_params['min_samples_leaf'],
                                       n_estimators=best_params['n_estimators'])
    
    random_forest_model.fit(X_train, y_train)

    models[dataset_name] = random_forest_model
```

#### Evaluation on Test Data

##### Evaluation of the Model Trained with 'No_Range_4': (X_No_Range_4_train_scaled, y_No_Range_4_train)

```python
from sklearn.metrics import precision_score, recall_score, f1_score, accuracy_score, log_loss, confusion_matrix, roc_auc_score

y_No_Range_4_pred = models['No_Range_4'].predict(X_No_Range_4_test_scaled)
y_No_Range_4_pred_proba = models['No_Range_4'].predict_proba(X_No_Range_4_test_scaled)

# Precision
precision_No_Range_4 = precision_score(y_No_Range_4_test, y_No_Range_4_pred, average='weighted')

# Recall
recall_No_Range_4 = recall_score(y_No_Range_4_test, y_No_Range_4_pred, average='weighted')

# F1-Score
f1_No_Range_4 = f1_score(y_No_Range_4_test, y_No_Range_4_pred, average='weighted')

# Classification Accuracy
accuracy_No_Range_4 = accuracy_score(y_No_Range_4_test, y_No_Range_4_pred)

# Log Loss
#log_loss_value_No_Range_4 = log_loss(y_No_Range_4_test, y_No_Range_4_pred)
log_loss_value_No_Range_4 = log_loss(y_No_Range_4_test, y_No_Range_4_pred_proba, labels=models['No_Range_4'].classes_)

# Confusion Matrix
conf_matrix_No_Range_4 = confusion_matrix(y_No_Range_4_test, y_No_Range_4_pred)

# AROC Curve
roc_auc_No_Range_4 = roc_auc_score(y_No_Range_4_test, y_No_Range_4_pred_proba, multi_class='ovr', labels=random_forest_model.classes_)

# Metrics
print(f'Test Confusion Matrix without No_Range_4:\n{conf_matrix_No_Range_4}')
print()
print(f'Test Area Under the ROC Curve (AROC) without No_Range_4: {roc_auc_No_Range_4}')
print(f'Test Accuracy without No_Range_4: {accuracy_No_Range_4}')
print(f'Test F1-Score without No_Range_4: {f1_No_Range_4}')
print(f'Test Precision without No_Range_4: {precision_No_Range_4}')
print(f'Test Recall without No_Range_4: {recall_No_Range_4}')
print(f'Training Log Loss without No_Range_4: {log_loss_value_No_Range_4}')
```

##### Evaluation of the Model Trained with 'No_Range': (X_No_Range_train_scaled, y_No_Range_train)

```python
y_No_Range_pred = models['No_Range'].predict(X_No_Range_test_scaled)
y_No_Range_pred_proba = models['No_Range'].predict_proba(X_No_Range_test_scaled)

# Precision
precision_No_Range = precision_score(y_No_Range_test, y_No_Range_pred, average='weighted')

# Recall
recall_No_Range = recall_score(y_No_Range_test, y_No_Range_pred, average='weighted')

# F1-Score
f1_No_Range = f1_score(y_No_Range_test, y_No_Range_pred, average='weighted')

# Classification Accuracy
accuracy_No_Range = accuracy_score(y_No_Range_test, y_No_Range_pred)

# Log Loss
log_loss_value_No_Range = log_loss(y_No_Range_test, y_No_Range_pred_proba, labels=models['No_Range'].classes_)

# Confusion Matrix
conf_matrix_No_Range = confusion_matrix(y_No_Range_test, y_No_Range_pred)

# AROC Curve
roc_auc_No_Range = roc_auc_score(y_No_Range_test, y_No_Range_pred_proba, multi_class='ovr', labels=models['No_Range'].classes_)

# Metrics
print(f'Test Confusion Matrix for No_Range:\n{conf_matrix_No_Range}')
print()
print(f'Test Area Under the ROC Curve (AROC) for No_Range: {roc_auc_No_Range}')
print(f'Test Accuracy for No_Range: {accuracy_No_Range}')
print(f'Test F1-Score for No_Range: {f1_No_Range}')
print(f'Test Precision for No_Range: {precision_No_Range}')
print(f'Test Recall for No_Range: {recall_No_Range}')
print(f'Training Log Loss for No_Range: {log_loss_value_No_Range}')
```

##### Area under the Receiver Operating Characteristic (AROC) Curve

```python
from sklearn.metrics import roc_curve, auc
from sklearn.preprocessing import LabelBinarizer

plt.figure(figsize=(7, 5))

model_names = ['No_Range_4', 'No_Range', 'Ranges_4', 'random_forest']
datasets = [X_No_Range_4_test_scaled, X_No_Range_test_scaled, X_Ranges_4_scaled]
labels = [y_No_Range_4_test, y_No_Range_test, y_Ranges_4]
model_colors = ['red', 'green', 'cyan']

for model_name, dataset, label, color in zip(model_names, datasets, labels, model_colors):
    y_pred_proba = models[model_name].predict_proba(dataset)
    
    label_binarizer = LabelBinarizer()
    y_binary = label_binarizer.fit_transform(label)
    
    fpr, tpr, _ = roc_curve(y_binary.ravel(), y_pred_proba.ravel())
    roc_auc = auc(fpr, tpr)

    plt.plot(fpr, tpr, label=f'{model_name} (Avg AUC = {roc_auc:.2f})', linestyle='-', linewidth=2, color=color)

plt.plot([0, 1], [0, 1], color='black', lw=2, linestyle='--', label='Random Classifier (AUC = 0.5)')

plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Average ROC Curves for Different Models')
plt.legend(loc='lower right')

plt.grid(True)
plt.tight_layout()
plt.show()
```
