#### Load the Dataset

```python
import pandas as pd
import numpy as np

path_train = "/home/sarantis/Documents/Διπλωματική Εργασία/Diploma-Thesis/PV_dataset/train data.csv"  
train_data = pd.read_csv(path_train, delimiter=';')

path_test = "/home/sarantis/Documents/Διπλωματική Εργασία/Diploma-Thesis/PV_dataset/test data.csv"  
test_data = pd.read_csv(path_test, delimiter=';')

display(train_data)
display(test_data)
```

#### Create new Datasets without some attributes

```python
# Dataframe without Range 4
Range4_train = train_data.drop('range 4', axis=1) 
Range4_test = test_data.drop('range 4', axis=1) 

# Dataframe without Range 1, Range 2, Range 3, Range 4
Ranges_train = train_data.drop(['range 1', 'range 2', 'range 3', 'range 4'], axis=1)
Ranges_test = test_data.drop(['range 1', 'range 2', 'range 3', 'range 4'], axis=1)

# Dataframe μόνο με τα 4 Ranges
I_train = train_data.drop(['I1', 'I2', 'I1MAX', 'I1MIN', 'I1VAR', 'I2MAX', 'I2MIN', 'I2VAR', 'I3', 'I4', 'I3max', 'I3min', 'I3var', 'I4MAX', 'I4MIN', 'I6', 'I5', 'Itotal1', 'Itotalmax1', 'Itotalmin1', 'Vdcmean1', 'Vdcmax1', 'Vdcmin1', 'Pdcmean1', 'IR', 'T'], axis=1)
I_test = test_data.drop(['I1', 'I2', 'I1MAX', 'I1MIN', 'I1VAR', 'I2MAX', 'I2MIN', 'I2VAR', 'I3', 'I4', 'I3max', 'I3min', 'I3var', 'I4MAX', 'I4MIN', 'I6', 'I5', 'Itotal1', 'Itotalmax1', 'Itotalmin1', 'Vdcmean1', 'Vdcmax1', 'Vdcmin1', 'Pdcmean1', 'IR', 'T'], axis=1)
```

```python
# Dataframes without class
X_Range4_train = Range4_train.drop('class', axis=1) 
X_Range4_test = Range4_test.drop('class', axis=1)

X_Ranges_train = Ranges_train.drop('class', axis=1) 
X_Ranges_test = Ranges_test.drop('class', axis=1)

X_I_train = I_train.drop('class', axis=1) 
X_I_test = I_test.drop('class', axis=1)

# Dataframe for Target Variable
y_Range4_train = Range4_train['class']  
y_Range4_test = Range4_test['class']

y_Ranges_train = Ranges_train['class']  
y_Ranges_test = Ranges_test['class']

y_I_train = I_train['class']  
y_I_test = I_test['class']
```

#### Min-Max Scaling

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

X_Range4_train_scaled = pd.DataFrame(scaler.fit_transform(X_Range4_train), columns=X_Range4_train.columns)
X_Range4_test_scaled = pd.DataFrame(scaler.transform(X_Range4_test), columns=X_Range4_test.columns)

X_Ranges_train_scaled = pd.DataFrame(scaler.fit_transform(X_Ranges_train), columns=X_Ranges_train.columns)
X_Ranges_test_scaled = pd.DataFrame(scaler.transform(X_Ranges_test), columns=X_Ranges_test.columns)

X_I_train_scaled = pd.DataFrame(scaler.fit_transform(X_I_train), columns=X_I_train.columns)
X_I_test_scaled = pd.DataFrame(scaler.transform(X_I_test), columns=X_I_test.columns)
```

#### Random Forest Algorithm

```python
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestClassifier

datasets = {
    'Range4': (X_Range4_train_scaled, y_Range4_train),
    'Ranges': (X_Ranges_train_scaled, y_Ranges_train),
    'I': (X_I_train_scaled, y_I_train)
}

param_grid = {
    'max_depth': [2, 3, 5, 10, 20],
    'min_samples_leaf': [5, 10, 20, 50, 100, 200],
    'n_estimators': [10, 25, 30, 50, 100, 200]
}

# Dictionairy to save the trained models
models = {}

random_forest = RandomForestClassifier(random_state=42)

for dataset_name, (X_train, y_train) in datasets.items():
    grid_search = GridSearchCV(estimator=random_forest, param_grid=param_grid, cv=5, scoring='accuracy')
    grid_search.fit(X_train, y_train)

    best_params = grid_search.best_params_
    best_score = grid_search.best_score_

    print(f'Best Parameters for {dataset_name}: {best_params}')
    print(f'Best Mean Cross-Validated Score for {dataset_name}: {best_score}')

    random_forest_model = RandomForestClassifier(random_state=42, max_depth=best_params['max_depth'],
                                       min_samples_leaf=best_params['min_samples_leaf'],
                                       n_estimators=best_params['n_estimators'])
    
    random_forest_model.fit(X_train, y_train)

    models[dataset_name] = random_forest_model
```

#### Evaluation on Test Data

##### Evaluation of the Model Trained with 'Range4': (X_Range4_train_scaled, y_Range4_train)

```python
from sklearn.metrics import precision_score, recall_score, f1_score, accuracy_score, log_loss, confusion_matrix, roc_auc_score

y_Range4_pred = models['Range4'].predict(X_Range4_test_scaled)
y_Range4_pred_proba = models['Range4'].predict_proba(X_Range4_test_scaled)

# Precision
precision_Range4 = precision_score(y_Range4_test, y_Range4_pred, average='weighted')

# Recall
recall_Range4 = recall_score(y_Range4_test, y_Range4_pred, average='weighted')

# F1-Score
f1_Range4 = f1_score(y_Range4_test, y_Range4_pred, average='weighted')

# Classification Accuracy
accuracy_Range4 = accuracy_score(y_Range4_test, y_Range4_pred)

# Log Loss
#log_loss_value_Range4 = log_loss(y_Range4_test, y_Range4_pred)
log_loss_value_Range4 = log_loss(y_Range4_test, y_Range4_pred_proba, labels=models['Range4'].classes_)

# Confusion Matrix
conf_matrix_Range4 = confusion_matrix(y_Range4_test, y_Range4_pred)

# AROC Curve
roc_auc_Range4 = roc_auc_score(y_Range4_test, y_Range4_pred_proba, multi_class='ovr', labels=random_forest_model.classes_)

# Metrics
print(f'Test Confusion Matrix without Range4:\n{conf_matrix_Range4}')
print()
print(f'Test Area Under the ROC Curve (AROC) without Range4: {roc_auc_Range4}')
print(f'Test Accuracy without Range4: {accuracy_Range4}')
print(f'Test F1-Score without Range4: {f1_Range4}')
print(f'Test Precision without Range4: {precision_Range4}')
print(f'Test Recall without Range4: {recall_Range4}')
print(f'Training Log Loss without Range4: {log_loss_value_Range4}')
```

##### Evaluation of the Model Trained with 'Ranges': (X_Ranges_train_scaled, y_Ranges_train)

```python
y_Ranges_pred = models['Ranges'].predict(X_Ranges_test_scaled)
y_Ranges_pred_proba = models['Ranges'].predict_proba(X_Ranges_test_scaled)

# Precision
precision_Ranges = precision_score(y_Ranges_test, y_Ranges_pred, average='weighted')

# Recall
recall_Ranges = recall_score(y_Ranges_test, y_Ranges_pred, average='weighted')

# F1-Score
f1_Ranges = f1_score(y_Ranges_test, y_Ranges_pred, average='weighted')

# Classification Accuracy
accuracy_Ranges = accuracy_score(y_Ranges_test, y_Ranges_pred)

# Log Loss
log_loss_value_Ranges = log_loss(y_Ranges_test, y_Ranges_pred_proba, labels=models['Ranges'].classes_)

# Confusion Matrix
conf_matrix_Ranges = confusion_matrix(y_Ranges_test, y_Ranges_pred)

# AROC Curve
roc_auc_Ranges = roc_auc_score(y_Ranges_test, y_Ranges_pred_proba, multi_class='ovr', labels=models['Ranges'].classes_)

# Metrics
print(f'Test Confusion Matrix for Ranges:\n{conf_matrix_Ranges}')
print()
print(f'Test Area Under the ROC Curve (AROC) for Ranges: {roc_auc_Ranges}')
print(f'Test Accuracy for Ranges: {accuracy_Ranges}')
print(f'Test F1-Score for Ranges: {f1_Ranges}')
print(f'Test Precision for Ranges: {precision_Ranges}')
print(f'Test Recall for Ranges: {recall_Ranges}')
print(f'Training Log Loss for Ranges: {log_loss_value_Ranges}')
```

##### Evaluation of the Model Trained with 'I': (X_I_train_scaled, y_I_train)

```python
y_I_pred = models['I'].predict(X_I_test_scaled)
y_I_pred_proba = models['I'].predict_proba(X_I_test_scaled)

# Precision
precision_I = precision_score(y_I_test, y_I_pred, average='weighted')

# Recall
recall_I = recall_score(y_I_test, y_I_pred, average='weighted')

# F1-Score
f1_I = f1_score(y_I_test, y_I_pred, average='weighted')

# Classification Accuracy
accuracy_I = accuracy_score(y_I_test, y_I_pred)

# Log Loss
log_loss_value_I = log_loss(y_I_test, y_I_pred_proba, labels=models['I'].classes_)

# Confusion Matrix
conf_matrix_I = confusion_matrix(y_I_test, y_I_pred)

# AROC Curve
roc_auc_I = roc_auc_score(y_I_test, y_I_pred_proba, multi_class='ovr', labels=models['I'].classes_)

# Metrics
print(f'Test Confusion Matrix for I:\n{conf_matrix_I}')
print()
print(f'Test Area Under the ROC Curve (AROC) for I: {roc_auc_I}')
print(f'Test Accuracy for I: {accuracy_I}')
print(f'Test F1-Score for I: {f1_I}')
print(f'Test Precision for I: {precision_I}')
print(f'Test Recall for I: {recall_I}')
print(f'Training Log Loss for I: {log_loss_value_I}')
```

##### Area under the Receiver Operating Characteristic (AROC) Curve

```python
from sklearn.metrics import roc_curve, auc
from sklearn.preprocessing import LabelBinarizer
import matplotlib.pyplot as plt

plt.figure(figsize=(7, 5))

model_names = ['Range4', 'Ranges', 'I']
datasets = [X_Range4_test_scaled, X_Ranges_test_scaled, X_I_test_scaled]
labels = [y_Range4_test, y_Ranges_test, y_I_test]
model_colors = ['red', 'green', 'cyan']

for model_name, dataset, label, color in zip(model_names, datasets, labels, model_colors):
    y_pred_proba = models[model_name].predict_proba(dataset)
    
    label_binarizer = LabelBinarizer()
    y_binary = label_binarizer.fit_transform(label)
    
    fpr, tpr, _ = roc_curve(y_binary.ravel(), y_pred_proba.ravel())
    roc_auc = auc(fpr, tpr)

    plt.plot(fpr, tpr, label=f'{model_name} (Avg AUC = {roc_auc:.2f})', linestyle='-', linewidth=2, color=color)


plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')

# Curve for the Original Dataset
plt.axhline(y=1, color='black', linestyle=':', linewidth=2, label='Avg AUC = 1 (Original Dataset)')

plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Average ROC Curves for Different Models')
plt.legend(loc='lower right')

plt.grid(True)
plt.tight_layout()
plt.show()
```
