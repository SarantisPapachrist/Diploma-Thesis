#### Mean Squared Error

```python
%store -r mse_test_svm_new
%store -r mse_test_rf_new
%store -r mse_test_xgb_new
%store -r mse_test_knn_new
%store -r mse_test_new
%store -r mse_lstm
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

mse_values = {
    'Classifier': ['LSTM', 'Random Forest', 'XGBoost', 'KNN', 'Linear Regression', 'SVM'],
    'MSE': [mse_lstm, mse_test_rf_new, mse_test_xgb_new, mse_test_knn_new, mse_test_new, mse_test_svm_new]
}

mse_data = pd.DataFrame(mse_values)

sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 6))
sns.barplot(x='Classifier', y='MSE', hue='Classifier', data=mse_data, dodge=False)
plt.xlabel('Classifier')
plt.ylabel('Mean Squared Error (MSE)')
plt.title('Classifier Mean Squared Errors')

for index, row in mse_data.iterrows():
    plt.text(index, row['MSE'] + 0.01, f'{row["MSE"]:.2f}', color='black', ha="center")

plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
